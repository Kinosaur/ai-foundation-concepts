## 2.2 Assignment 2

**Instruction**
1. Create an **integer** tensor `X2` of shape `(4, 4)` with random integers in `[0..10)`.
2. Convert it to **float** and store in `X2_float`.
3. Show the **difference** `(X2_float - X2)` and explain why the difference is zero or non-zero.


```python
# Assignment 2
X2 = torch.randint(low=0, high=10, size=(4, 4), dtype=torch.int64)
X2_float = X2.float()

diff = X2_float - X2
print("X2:", X2)
print("X2_float:", X2_float)
print("Difference (X2_float - X2):", diff)

X2_2 = X2_float.clone()
X2_2[0][0] = X2_2[0][0] + 0.5
diff2 = X2_2 - X2
print("Modified X2_2:", X2_2)
print("Difference after modification (X2_2 - X2):", diff2)
diff2 = diff2.int()
print("Difference after conversion to int:", diff2)

# Explanation:
# Converting int -> float preserves the exact integer values (for small ints like 0..9),
# so the difference is typically all zeros.
```

## 2.3 Assignment 3

**Instruction**
1. Create a random **float** tensor `X3` of shape `(3, 2)`.
2. **Transpose** it to get shape `(2, 3)`.
3. Perform **matrix multiplication** of `X3 @ X3_transposed`.
4. Print the resulting **shape**.


```python
# Assignment 3
X3 = torch.randn(3, 2)
X3_transposed = X3.T  # shape (2, 3)

result = X3 @ X3_transposed
print("X3 shape:", X3.shape)
print("X3_transposed shape:", X3_transposed.shape)
print("Result shape:", result.shape)  # expected (3, 3)
```

## 2.7 Assignment 7

**Instruction**
1. Create two random float tensors `X7_A` and `X7_B` each of shape `(2, 2)`.
2. **Concatenate** them along dimension `0` ⇒ result shape `(4, 2)`.
3. **Concatenate** them along dimension `1` ⇒ result shape `(2, 4)`.
4. Print both results.


```python
# Assignment 7
X7_A = torch.randn(2, 2)
X7_B = torch.randn(2, 2)

cat_dim0 = torch.cat([X7_A, X7_B], dim=0)  # (4,2)
cat_dim1 = torch.cat([X7_A, X7_B], dim=1)  # (2,4)

print("X7_A:", X7_A)
print("X7_B:", X7_B)

print("\nConcatenate along dim=0 (shape {}):\n{}".format(tuple(cat_dim0.shape), cat_dim0))
print("\nConcatenate along dim=1 (shape {}):\n{}".format(tuple(cat_dim1.shape), cat_dim1))
```

## 2.8 Assignment 8

**Instruction**

2. Compute the **maximum value** in the entire tensor.

4. Print the shapes of the resulting tensors.

## 2.9 Assignment 9

**Instruction**
1. Create a random float tensor `X9` of shape `(3, 3)`.
2. Compute the **determinant** of this matrix using `torch.linalg.det(X9)` (PyTorch 1.10+).
3. Print the result and interpret it (could be near zero if the matrix is nearly singular).


```python
# Assignment 9
X9 = torch.randn(3, 3)
det_X9 = torch.linalg.det(X9)

print("X9:", X9)
print("\nDeterminant:", det_X9.item())

# Interpretation (basic):
# - If det is near 0, the matrix is close to singular (not invertible or numerically unstable to invert).
# - If det is far from 0, it's generally well-conditioned for inversion (not guaranteed, but a useful hint).
```
